# Pipeline Walkthrough — main.py
**Every function call in `run_pipeline()` broken open, visualised step by step.**

```python
def run_pipeline(image_path):
    original_bgr     = cv2.imread(image_path)           # Step 1
    gray             = preprocess_pipeline(image_path)  # Steps 2-4
    h, w             = gray.shape
    original_resized = cv2.resize(original_bgr, (w, h)) # Step 5
    model            = CustomCNN()                      # Step 6
    mask             = segment_image(gray, original_resized, model)  # Steps 7-11
    result           = extract_boundary(original_resized, mask)      # Steps 12-13
    return original_resized, result
```

````python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import os, sys

sys.path.append(os.path.abspath('..'))

from utils.preprocess import preprocess_pipeline
from cnn.cnn_model    import CustomCNN
from cnn.conv         import convolve
from cnn.activation   import relu
from features.statistics import compute_stats
from segmentation.patch_segmenter import segment_image
from main import extract_boundary, run_pipeline

plt.rcParams.update({'figure.dpi': 120, 'axes.titleweight': 'bold', 'axes.titlesize': 10})

IMAGE_PATH = os.path.join('..', 'data', 'testcases', '1.png')
print('Image:', IMAGE_PATH)
````

---
## Step 1 — `original_bgr = cv2.imread(image_path)`

````python
original_bgr = cv2.imread(IMAGE_PATH)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
fig.suptitle('Step 1 — cv2.imread()  →  raw BGR numpy array', fontweight='bold')

axes[0].imshow(cv2.cvtColor(original_bgr, cv2.COLOR_BGR2RGB))
axes[0].set_title(f'original_bgr  shape={original_bgr.shape}  dtype={original_bgr.dtype}')
axes[0].axis('off')

for c, (col, lab) in enumerate(zip(['blue','green','red'], ['B ch','G ch','R ch'])):
    axes[1].hist(original_bgr[:,:,c].flatten(), bins=60, color=col, alpha=0.5, label=lab)
axes[1].set_title('BGR channel distributions  (note: OpenCV loads as BGR, not RGB)')
axes[1].set_xlabel('Pixel value  [0–255]')
axes[1].legend()

plt.tight_layout()
plt.show()
print(f'Shape  : {original_bgr.shape}   → (H, W, 3)')
print(f'dtype  : {original_bgr.dtype}')
print(f'Range  : [{original_bgr.min()}, {original_bgr.max()}]')
print(f'Sample pixel [50,50] BGR: {original_bgr[50,50]}')
````

---
## Steps 2–4 — `gray = preprocess_pipeline(image_path)`
Three sub-steps inside this single call: grayscale → resize → normalise.

````python
# ── Manually break open preprocess_pipeline() ─────────────────────────────
from utils.preprocess import load_image, to_grayscale, resize_image, normalize_image

_img   = load_image(IMAGE_PATH)         # same as cv2.imread
step2  = to_grayscale(_img)             # BGR → single channel
step3  = resize_image(step2, (256,256)) # resize
step4  = normalize_image(step3)         # /255 → float32
gray   = step4                          # this is what run_pipeline() gets

stages = [
    ('Step 2\nto_grayscale()',   step2,  'gray',   f'shape={step2.shape}\ndtype={step2.dtype}'),
    ('Step 3\nresize_image()',   step3,  'gray',   f'shape={step3.shape}\ndtype={step3.dtype}'),
    ('Step 4\nnormalize_image()',step4,  'gray',   f'shape={step4.shape}\ndtype={step4.dtype}\nrange=[{step4.min():.2f}, {step4.max():.2f}]'),
]

fig, axes = plt.subplots(2, 3, figsize=(14, 7))
fig.suptitle('Steps 2–4 — preprocess_pipeline() broken open', fontweight='bold')

for col, (title, img, cmap, info) in enumerate(stages):
    axes[0, col].imshow(img, cmap=cmap)
    axes[0, col].set_title(f'{title}\n{info}')
    axes[0, col].axis('off')
    axes[1, col].hist(img.flatten(), bins=50, color='steelblue', alpha=0.8)
    axes[1, col].set_xlabel('Pixel value')
    axes[1, col].set_title('Distribution')

plt.tight_layout()
plt.show()
print(f'Final gray  shape={gray.shape}  dtype={gray.dtype}  range=[{gray.min():.3f}, {gray.max():.3f}]')
````

---
## Step 5 — `original_resized = cv2.resize(original_bgr, (w, h))`

````python
h, w = gray.shape              # 256, 256
original_resized = cv2.resize(original_bgr, (w, h))

fig, axes = plt.subplots(1, 2, figsize=(10, 4))
fig.suptitle('Step 5 — Resize colour image to match gray dimensions', fontweight='bold')

axes[0].imshow(cv2.cvtColor(original_bgr, cv2.COLOR_BGR2RGB))
axes[0].set_title(f'original_bgr  {original_bgr.shape[1]}×{original_bgr.shape[0]}')
axes[0].axis('off')

axes[1].imshow(cv2.cvtColor(original_resized, cv2.COLOR_BGR2RGB))
axes[1].set_title(f'original_resized  {original_resized.shape[1]}×{original_resized.shape[0]}')
axes[1].axis('off')

plt.tight_layout()
plt.show()
print(f'Why? segment_image() needs both arrays to have the SAME (h,w) so patch slicing is aligned.')
print(f'gray.shape = {gray.shape}  |  original_resized.shape = {original_resized.shape[:2]}')
````

---
## Step 6 — `model = CustomCNN()`
No computation happens here — just initialises the 4 kernels.

````python
model = CustomCNN()

names = ['Sobel-X\n(vertical edges)', 'Sobel-Y\n(horizontal edges)',
         'Laplacian\n(blobs/corners)', 'Diagonal\n(diagonal contrast)']

fig, axes = plt.subplots(1, 4, figsize=(14, 3.5))
fig.suptitle('Step 6 — CustomCNN() — the 4 fixed kernels (no training, no weights)', fontweight='bold')

for ax, kernel, name in zip(axes, model.kernels, names):
    im = ax.imshow(kernel, cmap='RdBu_r', vmin=-4, vmax=4)
    for r in range(3):
        for c in range(3):
            ax.text(c, r, f'{int(kernel[r,c]):+d}', ha='center', va='center',
                    fontsize=14, fontweight='bold',
                    color='white' if abs(kernel[r,c]) > 1 else 'black')
    ax.set_title(name)
    ax.set_xticks([]); ax.set_yticks([])
    plt.colorbar(im, ax=ax, fraction=0.046)

plt.tight_layout()
plt.show()
print(f'model.kernels is a list of {len(model.kernels)} numpy arrays, each shape (3,3)')
````

---
## Steps 7–11 — `mask = segment_image(gray, original_resized, model)`
This is the main function. We break it into 5 sub-steps.

### Step 7 — Sliding window: extract one patch, run CNN, compute score

````python
PATCH_SIZE = 16
stride     = PATCH_SIZE // 4    # = 4
R, C       = 100, 100           # one example patch position

patch_gray  = gray[R:R+PATCH_SIZE, C:C+PATCH_SIZE]
patch_color = original_resized[R:R+PATCH_SIZE, C:C+PATCH_SIZE]

# — CNN forward pass on this one patch —
feature_maps         = model.forward(patch_gray)
mean_val, var_val, _ = compute_stats(feature_maps)

# — Three scoring cues —
smoothness    = 1.0 / (var_val + 1e-6)
B_mean        = float(patch_color[:,:,0].mean())
R_mean        = float(patch_color[:,:,2].mean())
blue_score    = max(0.0, B_mean - R_mean)
gx = cv2.Sobel(patch_gray, cv2.CV_32F, 1, 0, ksize=3)
gy = cv2.Sobel(patch_gray, cv2.CV_32F, 0, 1, ksize=3)
gradient_energy = float(np.mean(np.sqrt(gx**2 + gy**2)))
water_score   = 0.6*smoothness + 0.3*blue_score - 0.5*gradient_energy

fig, axes = plt.subplots(1, 6, figsize=(20, 4))
fig.suptitle(f'Step 7 — One patch at (row={R}, col={C}): CNN → stats → water score', fontweight='bold')

axes[0].imshow(patch_gray, cmap='gray', vmin=0, vmax=1)
axes[0].set_title('Grayscale patch\n(16×16)'); axes[0].axis('off')

axes[1].imshow(cv2.cvtColor(patch_color, cv2.COLOR_BGR2RGB))
axes[1].set_title('Colour patch\n(16×16)'); axes[1].axis('off')

for i, (fm, nm) in enumerate(zip(feature_maps, ['SobelX','SobelY','Lapl','Diag'])):
    axes[2].imshow(feature_maps[0], cmap='hot')
axes[2].set_title(f'Feature map 0\n(Sobel-X |relu|)'); axes[2].axis('off')

axes[3].bar(['var'], [var_val], color='steelblue')
axes[3].set_title(f'CNN variance\n= {var_val:.5f}')

components = ['0.6×smooth', '0.3×blue', '-0.5×grad']
vals = [0.6*smoothness, 0.3*blue_score, -0.5*gradient_energy]
axes[4].bar(components, vals, color=['#2980b9','#1a6ca8','#c0392b'])
axes[4].axhline(0, color='black', lw=1)
axes[4].set_title('Score components')
axes[4].tick_params(axis='x', labelsize=8)

axes[5].axis('off')
axes[5].text(0.1, 0.95,
    f'water_score =\n'
    f'  0.6 × {smoothness:.2f}\n'
    f'+ 0.3 × {blue_score:.2f}\n'
    f'- 0.5 × {gradient_energy:.3f}\n'
    f'─────────────\n'
    f'= {water_score:.4f}\n\n'
    f'→  {"WATER" if water_score>0 else "LAND"}',
    transform=axes[5].transAxes, va='top', fontfamily='monospace', fontsize=10,
    bbox=dict(boxstyle='round', facecolor='#e8f8e8'))

plt.tight_layout()
plt.show()
````

### Step 8 — All patches → accumulated score map

````python
h_img, w_img = gray.shape
water_score_map = np.zeros((h_img, w_img), dtype=np.float32)

patch_count = 0
for i in range(0, h_img - PATCH_SIZE + 1, stride):
    for j in range(0, w_img - PATCH_SIZE + 1, stride):
        pg = gray[i:i+PATCH_SIZE, j:j+PATCH_SIZE]
        pc = original_resized[i:i+PATCH_SIZE, j:j+PATCH_SIZE]
        _, var_v, _ = compute_stats(model.forward(pg))
        sm = 1.0/(var_v+1e-6)
        bs = max(0, float(pc[:,:,0].mean())-float(pc[:,:,2].mean()))
        gx2 = cv2.Sobel(pg, cv2.CV_32F, 1, 0, ksize=3)
        gy2 = cv2.Sobel(pg, cv2.CV_32F, 0, 1, ksize=3)
        ge = float(np.mean(np.sqrt(gx2**2+gy2**2)))
        water_score_map[i:i+PATCH_SIZE, j:j+PATCH_SIZE] += 0.6*sm + 0.3*bs - 0.5*ge
        patch_count += 1

score_map_norm = cv2.normalize(water_score_map, None, 0, 1, cv2.NORM_MINMAX)

fig, axes = plt.subplots(1, 3, figsize=(14, 4))
fig.suptitle(f'Step 8 — {patch_count} patches accumulated → score map normalised to [0,1]', fontweight='bold')

axes[0].imshow(cv2.cvtColor(original_resized, cv2.COLOR_BGR2RGB))
axes[0].set_title('Input colour image'); axes[0].axis('off')

im = axes[1].imshow(score_map_norm, cmap='RdYlGn', vmin=0, vmax=1)
plt.colorbar(im, ax=axes[1], fraction=0.046)
axes[1].set_title('Normalised score map\nGreen=water  Red=land'); axes[1].axis('off')

axes[2].hist(score_map_norm.flatten(), bins=50, color='steelblue')
axes[2].axvline(0.5, color='red', linestyle='--', lw=2, label='threshold=0.5')
axes[2].set_title('Score distribution'); axes[2].set_xlabel('Score'); axes[2].legend()

plt.tight_layout()
plt.show()
````

### Step 9 — Threshold → binary mask

````python
_, mask_bin = cv2.threshold(score_map_norm, 0.5, 1, cv2.THRESH_BINARY)
mask_bin = (mask_bin * 255).astype(np.uint8)

overlay = cv2.cvtColor(original_resized, cv2.COLOR_BGR2RGB).copy().astype(float)
overlay[mask_bin>0] = overlay[mask_bin>0]*0.45 + np.array([0,60,200])*0.55

fig, axes = plt.subplots(1, 3, figsize=(14, 4))
fig.suptitle('Step 9 — Threshold at 0.5 → binary mask', fontweight='bold')

axes[0].imshow(score_map_norm, cmap='RdYlGn', vmin=0, vmax=1)
axes[0].set_title('Score map (continuous)'); axes[0].axis('off')

axes[1].imshow(mask_bin, cmap='gray')
axes[1].set_title(f'Binary mask\nWhite=water ({(mask_bin>0).sum()}px)  Black=land'); axes[1].axis('off')

axes[2].imshow(overlay.astype(np.uint8))
axes[2].set_title('Water region (blue tint)'); axes[2].axis('off')

plt.tight_layout()
plt.show()
````

### Step 10 — Keep largest connected component

````python
num_labels, labels = cv2.connectedComponents(mask_bin)
areas = sorted([(l, int(np.sum(labels==l))) for l in range(1,num_labels)],
               key=lambda x: x[1], reverse=True)

np.random.seed(42)
label_rgb = np.zeros((*labels.shape, 3), dtype=np.uint8)
for lbl, area in areas:
    label_rgb[labels==lbl] = np.random.randint(60, 255, 3)

clean_mask = np.zeros_like(mask_bin)
if areas:
    clean_mask[labels==areas[0][0]] = 255

fig, axes = plt.subplots(1, 3, figsize=(14, 4))
fig.suptitle('Step 10 — Keep largest connected component', fontweight='bold')

axes[0].imshow(mask_bin, cmap='gray')
axes[0].set_title(f'Binary mask\n{num_labels-1} component(s)'); axes[0].axis('off')

axes[1].imshow(label_rgb)
for lbl, area in areas[:5]:
    pos = np.argwhere(labels==lbl)
    cy, cx = pos.mean(axis=0)
    axes[1].text(cx, cy, f'{area}px', fontsize=8, ha='center', color='white', fontweight='bold')
axes[1].set_title('All components (unique colours)'); axes[1].axis('off')

axes[2].imshow(clean_mask, cmap='gray')
axes[2].set_title(f'Largest only\n({(clean_mask>0).sum()} px kept)'); axes[2].axis('off')

plt.tight_layout()
plt.show()
````

### Step 11 — Morphological Close + Open

````python
k = np.ones((7,7), np.uint8)
after_close = cv2.morphologyEx(clean_mask, cv2.MORPH_CLOSE, k)
mask        = cv2.morphologyEx(after_close, cv2.MORPH_OPEN,  k)   # ← this is the final mask

holes_filled  = ((after_close>0) & (clean_mask==0)).astype(np.uint8)*255
noise_removed = ((clean_mask>0)  & (mask==0)).astype(np.uint8)*255

fig, axes = plt.subplots(1, 4, figsize=(18, 4))
fig.suptitle('Step 11 — Morphological Close then Open  (7×7 kernel)', fontweight='bold')

axes[0].imshow(clean_mask, cmap='gray'); axes[0].set_title('Before'); axes[0].axis('off')
axes[1].imshow(after_close, cmap='gray'); axes[1].set_title('After CLOSE\n(holes filled)'); axes[1].axis('off')
axes[2].imshow(mask, cmap='gray'); axes[2].set_title('After OPEN\n(noise removed)  ← final mask'); axes[2].axis('off')

diff = np.stack([noise_removed, holes_filled, np.zeros_like(clean_mask)], axis=2)
axes[3].imshow(diff)
r_p = mpatches.Patch(color='red',   label=f'Removed ({(noise_removed>0).sum()}px)')
g_p = mpatches.Patch(color='green', label=f'Filled  ({(holes_filled>0).sum()}px)')
axes[3].legend(handles=[r_p,g_p], fontsize=9, loc='lower right')
axes[3].set_title('Changes'); axes[3].axis('off')

plt.tight_layout()
plt.show()
print(f'mask returned by segment_image()  →  shape={mask.shape}  dtype={mask.dtype}')
````

---
## Steps 12–13 — `result = extract_boundary(original_resized, mask)`

````python
# ── Step 12: findContours ───────────────────────────────────────────────────
contours, _ = cv2.findContours(mask.astype(np.uint8), cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

contour_canvas = np.zeros_like(mask)
cv2.drawContours(contour_canvas, contours, -1, 255, 2)

# ── Step 13: drawContours in red ────────────────────────────────────────────
output = original_resized.copy()
cv2.drawContours(output, contours, -1, (0, 0, 255), 2)
result = output

fig, axes = plt.subplots(1, 4, figsize=(18, 4))
fig.suptitle('Steps 12–13 — extract_boundary(): findContours → drawContours in red', fontweight='bold')

axes[0].imshow(mask, cmap='gray')
axes[0].set_title('Final mask  (input to\nextract_boundary())'); axes[0].axis('off')

axes[1].imshow(contour_canvas, cmap='gray')
axes[1].set_title(f'Extracted contours\n{len(contours)} contour(s)'); axes[1].axis('off')

axes[2].imshow(cv2.cvtColor(original_resized, cv2.COLOR_BGR2RGB))
axes[2].set_title('original_resized\n(colour input)'); axes[2].axis('off')

axes[3].imshow(cv2.cvtColor(result, cv2.COLOR_BGR2RGB))
axes[3].set_title('FINAL OUTPUT  result\n(red boundary overlay)'); axes[3].axis('off')

plt.tight_layout()
plt.show()
print(f'result  shape={result.shape}  dtype={result.dtype}')
````

---
## Return Values — `return original_resized, result`

````python
fig, axes = plt.subplots(1, 2, figsize=(10, 5))
fig.suptitle('run_pipeline() return values', fontweight='bold')

axes[0].imshow(cv2.cvtColor(original_resized, cv2.COLOR_BGR2RGB))
axes[0].set_title(f'original_resized\nshape={original_resized.shape}')
axes[0].axis('off')

axes[1].imshow(cv2.cvtColor(result, cv2.COLOR_BGR2RGB))
axes[1].set_title(f'result\nshape={result.shape}')
axes[1].axis('off')

plt.tight_layout()
plt.show()
````

---
## Full Strip — All Steps in One View

````python
steps = [
    ('1. imread',         cv2.cvtColor(original_bgr, cv2.COLOR_BGR2RGB), None),
    ('2. grayscale',      step2,         'gray'),
    ('3. resize 256²',    step3,         'gray'),
    ('4. normalise',      (step4*255).astype(np.uint8), 'gray'),
    ('5. color resize',   cv2.cvtColor(original_resized, cv2.COLOR_BGR2RGB), None),
    ('6. CNN kernels',    np.abs(relu(convolve(gray, model.kernels[0]))*255).astype(np.uint8), 'hot'),
    ('7. score (patch)',  (score_map_norm*255).astype(np.uint8), 'RdYlGn'),
    ('8. score map',      (score_map_norm*255).astype(np.uint8), 'RdYlGn'),
    ('9. threshold',      mask_bin,      'gray'),
    ('10. largest comp',  clean_mask,    'gray'),
    ('11. morphology',    mask,          'gray'),
    ('12. contours',      contour_canvas,'gray'),
    ('13. RESULT',        cv2.cvtColor(result, cv2.COLOR_BGR2RGB), None),
]

fig, axes = plt.subplots(1, 13, figsize=(30, 3.5))
fig.suptitle('main.py  run_pipeline()  —  Full Step-by-Step Strip', fontsize=13, fontweight='bold')

for ax, (title, img, cmap) in zip(axes, steps):
    kw = {'cmap': cmap} if cmap else {}
    if img.ndim == 2: kw.update({'vmin':0,'vmax':255})
    ax.imshow(img, **kw)
    ax.set_title(title, fontsize=7.5)
    ax.axis('off')

plt.tight_layout()
plt.show()
````
